```python
import numpy as np
import time
import math

# Create big data
N = 10_000_001
my_list = list(range(N))
arr = np.arange(N)

# Python loop version
t0 = time.perf_counter()
roots_list = [math.sqrt(num) for num in my_list]
t1 = time.perf_counter()

# NumPy version (vectorised)
roots_arr = np.sqrt(arr)
t2 = time.perf_counter()

print(f"Python list loop: {t1 - t0:.4f} seconds")
print(f"NumPy array vectorised loop: {t2 - t1:.4f} seconds")
```

NumPy Builtin Functions

```python
import numpy as np

#arange function
arr = np.arange(10)
print("Array with arange 10: ",arr)
arr = np.arange(1,10)
print("Array with arange 1 to 10: ",arr)
arr = np.arange(1,10,2)
print("Array with arange 1 to 10 with step 2: ",arr)
arr = np.arange(1,10,0.5)
print("Array with arange 1 to 10 with step 0.5: ",arr)

#zeros function
arr1z = np.zeros(shape=4) 
print("Array of 4 with Zero: ",arr1z)
arr2z = np.zeros(5) 
print("Array of 4 with Zero without shape kw: ",arr2z)
arr2d = np.zeros(shape=(2,5))
print("2D Array 2,5 with Zero: ",arr2d)
arr3d = np.zeros(shape=(2,2,5)) 
print("3D Array 2,2,5 with Zero: ",arr3d)

#onse function
arr = np.ones(shape=4)
print("Array of 4: ",arr)
arr2d = np.ones(shape=(2,5)) 
print("2D Array 2,5: ",arr2d)
arr3d = np.ones(shape=(2,2,5))
print("3D Array 2,2,5: ",arr3d)

#linspace function
arr = np.linspace(1,10,5)
print("Array with linspace 1 to 10 with 5 elements: ",arr)
arr = np.linspace(1,10,4)
print("Array with linspace 1 to 10 with 4 elements: ",arr)
arr = np.linspace(1,10,3)
print("Array with linspace 1 to 10 with 3 elements: ",arr)

#logspace function
arr = np.logspace(1,10,5)
print("Array with logspace 1 to 10 with 5 elements: ",arr)
```
