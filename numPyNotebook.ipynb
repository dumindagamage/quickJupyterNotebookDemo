```python
#makesure to install numpy package first with pip install numpy
 
import numpy as np

arr = np.array([1, 2, 3, 4, 5])
print("Original array:", arr)

print("Index:", arr[4])

print("Mask:", arr > 3)

print("Filtered array:", arr[arr > 3])
```

```python
#Play around with numpy arrays and indexing
import numpy as np
import math

grid = np.random.randint(1, 1_000, size=(5, 5))
print("Random grid:\n", grid)


#Array Operations
a = np.array([1, 2, 3, 4])
print(a * 2)  # Vectorized multiplication
print(a + 10) # broadcasting addition
print(np.sqrt(a))  # unoversal function

#without numpy
b = [1, 2, 3, 4]
print([num*2 for num in b ])
print([num+10 for num in b ])
print([math.sqrt(num) for num in b ])
```
